```python
%sql
create  or replace table  silver.cleaned_data.hotel_businesses
COMMENT "This table has SILVER data from the DCV feed"
TBLPROPERTIES ("quality" = "silver","delta.columnMapping.mode" = "name")
AS
select  
row_number() OVER ( ORDER BY place_id ASC) Hotelkey,
category hotel_category, 
address Hotel_address, 
open_hours Hotel_open_hours, 
reviews_count Hotel_reviews_count,
rating hotel_review_rating, 
main_image hotel_main_image, 
reviews hotel_reviews, 
place_id hotel_lace_id, 
lat hotel_latitude, 
lon hotel_longitude,
country hotel_country, 
services_provided hotel_services_provided , 
category_search_input hotel_category_search, 
open_website hotel_open_website, 
phone_number hotel_phone_number, 
hotel_star_ratings,
case when hotel_amenities like '%Pet-friendly%' then 'Yes' else '' end as hotel_is_pet_friendly
from raw.bright_initiative.google_maps_businesses_csv
where category like '%hotel%' or category like '%Hotel%'
```

```python
%sql
create  or replace table  silver.cleaned_data.vet_businesses
COMMENT "This table has SILVER data from the DCV feed"
TBLPROPERTIES ("quality" = "silver","delta.columnMapping.mode" = "name")
AS
select  
row_number() OVER ( ORDER BY place_id ASC) vetkey,
category vet_category, 
address vet_address, 
open_hours vet_open_hours,  
reviews_count vet_reviews_count,
rating vet_review_rating, 
main_image vet_main_image, 
reviews vet_reviews, 
place_id  vet_place_id, 
lat vet_latitude, 
lon vet_longitude, 
country ven_country, 
services_provided vet_services_provided, 
category_search_input vet_category_search, 
open_website vet_open_website, 
phone_number vet_phone_number, 
hotel_star_ratings vet_star_ratings
from raw.bright_initiative.google_maps_businesses_csv
where category = 'Animal hospital'
```
